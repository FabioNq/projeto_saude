## 2. Leitura dos arquivos

Este bloco carrega a base do Vigitel e a tabela auxiliar de unidades federativas/capitais:
- `pd.read_csv()` lê os arquivos CSV.
- `decimal=','` informa que a vírgula é usada como separador decimal na base do Vigitel.

## 4. Filtragem dos registros

Este bloco mantém em `df_vigitel` somente os registros que atendem simultaneamente a duas condições:
- `ano >= 2022`: registros a partir de 2022.
- `fet <= 3`: registros cujo valor da variável `fet` é menor ou igual a 3.

`fet` : representa a faixa etária : menor ou igual a 3 representa as faixa etaria entre 18 a 24 anos, 25 a 34 anos, 35 a 44 anos, que são as faixas etárias que queremos investigar  

O operador `&` combina as condições, de modo que ambas precisam ser verdadeiras para o registro ser mantido.


```python
df_vigitel = df_vigitel.loc[(df_vigitel['ano'] >= 2022) & (df_vigitel['fet'] <= 3)]
```

## 5. Verificação das dimensões da base filtrada

`df_vigitel.shape` retorna uma tupla com a quantidade de linhas e colunas do DataFrame após a filtragem. Assim, é possível verificar o tamanho da base resultante.

## 6. Seleção das colunas de interesse

A lista `colunas_especificas` define quais variáveis serão mantidas para a análise. Em seguida, `df_vigitel[colunas_especificas]` cria `df_vigitel_2` contendo apenas essas colunas.

O comando `df_vigitel_2.info()` apresenta a estrutura dessa seleção, incluindo tipos de dados e quantidade de valores não nulos por coluna.


```python
#Colunas
colunas_especificas = ['ano','cidade','pesorake2025','q6','q7','fet','q8a','q9','q11','imc_i','frutareg','refritl5','score_upp_2cat','score_upp_2cat2024','inativo_2023','ativo_livre','tv_d_3','tempo_tela_total','fumante','r403','r403a','fumantetotal','alcabu','depressao','hart','diab','obesid_i','excpeso_i','r701_hh','saruim']

df_vigitel_2 = df_vigitel[colunas_especificas]
df_vigitel_2.info()
```

## 8. Combinação de variáveis e exclusão das colunas originais

Este bloco combina pares de colunas que representam versões alternativas de uma mesma informação:
- `combine_first()` preenche os valores ausentes da primeira coluna com valores disponíveis na segunda.
- O resultado é armazenado em `score_upp_2catGeral` e `r403_Geral`.
- `isna().sum()` contabiliza os valores ausentes por coluna antes da exclusão.
- `drop(columns=...)` remove as quatro colunas originais que foram combinadas, mantendo as versões consolidadas.

O uso de `.copy()` cria uma cópia explícita do DataFrame antes das alterações.


```python
#Mesclando coluna score_upp_2cat com score_upp_2cat2024
df_vigitel_2 = df_vigitel_2.copy() 
df_vigitel_2['score_upp_2catGeral'] = df_vigitel_2['score_upp_2cat'].combine_first(df_vigitel_2['score_upp_2cat2024'])

df_vigitel_2['r403_Geral'] = df_vigitel_2['r403'].combine_first(df_vigitel_2['r403a'])


#Excluir variaveis mescladas
df_vigitel_2.isna().sum()
colunas_para_excluir = ['r403','r403a','score_upp_2cat','score_upp_2cat2024']
df_vigitel_2 = df_vigitel_2.drop(columns = colunas_para_excluir)
```

## 9. Conversão de tipos de dados

Este bloco converte as colunas `pesorake2025`, `q9` e `imc_i` para o tipo numérico `float`. Isso permite realizar cálculos matemáticos e análises estatísticas com essas variáveis, desde que seus valores possam ser convertidos para números.


```python
#CONVERTER Variaveis Object para float
df_vigitel_2 = df_vigitel_2.copy()
colunas = ['pesorake2025','q9','imc_i']
df_vigitel_2[colunas] = df_vigitel_2[colunas].astype(float)
```

## 10. Conferência dos tipos após a conversão

`df_vigitel_2.info()` mostra novamente a estrutura do DataFrame. A saída permite conferir se as colunas selecionadas passaram a ter os tipos esperados e verificar quantos valores não nulos existem em cada uma.


```python
df_vigitel_2.info()
```

## 11. Preenchimento de valores ausentes pela média

Este bloco calcula a média das colunas `q8a` e `r701_hh` e usa essas médias para preencher os valores ausentes (`NaN`) nas respectivas colunas.

A operação `isna()` identifica as posições ausentes, e `.loc[...]` seleciona essas posições para receber a média calculada. Esse procedimento altera os valores ausentes; a média pode não ser adequada para toda variável, dependendo do significado e da distribuição dos dados.


```python
#Atribui média a valores faltantes

media_coluna = df_vigitel_2['q8a'].mean()
media_coluna_h = df_vigitel_2['r701_hh'].mean()


df_vigitel_2.loc[df_vigitel_2['q8a'].isna() ,'q8a'] = media_coluna
df_vigitel_2.loc[df_vigitel_2['r701_hh'].isna() ,'r701_hh'] = media_coluna_h


```

## 12. Contagem de valores ausentes por coluna

`df_vigitel_2.isna().sum()` conta quantos valores ausentes existem em cada coluna após o preenchimento realizado no bloco anterior. A saída ajuda a identificar quais variáveis ainda contêm dados faltantes.


```python
#VERIFICAR DADOS FALTANTES por coluna
df_vigitel_2.isna().sum()
```

## 13. Tratamento de códigos especiais e preenchimento pela média

A lista `colunas_atribui_vazio` seleciona as colunas `q8a`, `q9` e `q11`.

Para essas colunas:
- `apply()` aplica uma operação a cada coluna.
- `np.where()` substitui os códigos `888` e `777` por `NaN`.
- Uma segunda aplicação de `apply()` usa `fillna(x.mean())` para preencher os valores ausentes com a média de cada coluna.

O código comentado não é executado. Antes de tratar códigos como `777` e `888` como ausentes, é importante confirmar no dicionário de dados que esses códigos realmente representam respostas especiais, e não valores válidos.


```python
#VERIFICAR DADOS FALTANTES por Linhas
colunas_atribui_vazio = ['q8a','q9','q11']

#df_vigitel_2[colunas_atribui_vazio] = df_vigitel_2[colunas_atribui_vazio].replace#((888|777),np.nan)

# O lambda avalia cada coluna (x) e substitui onde for 888 e 777
df_vigitel_2[colunas_atribui_vazio] = df_vigitel_2[colunas_atribui_vazio].apply(lambda x: np.where((x == 888) | (x == 777), np.nan, x))

# Preenche com a médias os valores NaN
df_vigitel_2[colunas_atribui_vazio] = df_vigitel_2[colunas_atribui_vazio].apply(lambda x: x.fillna(x.mean()))

```

## 14. Resumo estatístico após o tratamento

`df_vigitel_2.describe()` apresenta novamente as estatísticas descritivas depois das transformações e do tratamento de valores ausentes. A comparação com a saída anterior pode ajudar a observar como as alterações afetaram os dados.


```python
#Atribuir a media para os nulos 
df_vigitel_2.describe()

```

## 15. Junção com a tabela de capitais

Este bloco usa `merge()` para acrescentar a `df_vigitel_2` o nome da capital correspondente ao código presente na coluna `cidade`:
- Da tabela `df_uf`, são selecionadas as colunas `codDici` e `Capital`.
- `left_on='cidade'` e `right_on='codDici'` definem as colunas usadas para relacionar as tabelas.
- `how='left'` mantém todas as linhas da base do Vigitel, mesmo quando não há correspondência na tabela auxiliar.
- O resultado da junção é armazenado em `df_vigitel_Uf`.

A última linha, `df_vigitel_2`, exibe o DataFrame anterior à junção; para visualizar a tabela já enriquecida com o nome da capital, seria necessário exibir `df_vigitel_Uf`.


```python
#Juntando Tabela UF com a tabela Vigitel para ter o nome das cidades

df_vigitel_Uf = df_vigitel_2.merge(df_uf[['codDici','Capital']],left_on='cidade',right_on='codDici',how='left')


df_vigitel_2
```

Como o argumento `index` foi definido, o Pandas não inclui o índice do DataFrame como uma coluna adicional no arquivo exportado.
