```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
from scipy import stats
from scipy.stats import norm

```

```python
df = pd.read_csv('dados/dados_Vigitel_Analise.csv')
```

```python
# 2. Filtra apenas a faixa etária de 18 a 44 anos (variável q6 = idade)
df_18_44 = df[(df['q6'] >= 18) & (df['q6'] <= 44)].copy()

# 3. Mapeamento oficial dos códigos das 27 capitais no Vigitel
df_18_44 = df[(df['q6'] >= 18) & (df['q6'] <= 44)].copy()


# 4. Calcula o percentual simples (sem peso amostral) por capital
resultado = df_18_44.groupby('Capital', as_index=False).agg({
    'fumante': lambda x: round(x.mean() * 100, 1),
    'excpeso_i': lambda x: round(x.mean() * 100, 1),
    'obesid_i': lambda x: round(x.mean() * 100, 1),
    'ativo_livre': lambda x: round(x.mean() * 100, 1),
    'hart':lambda x: round(x.mean()*100,1),
    'depressao':lambda x: round(x.mean()*100,1),
    'saruim':lambda x: round(x.mean()*100,1),
    'alcabu':lambda x: round(x.mean()*100,1)
    
    
}).rename(columns={
    'fumante': 'Fumantes (%)',
    'excpeso_i': 'Excesso de Peso (%)',
    'obesid_i': 'Obesidade (%)',
    'ativo_livre': 'Ativ. Física Insuficiente (%)',
    'hart': 'problema no coracao %',
    'depressao': 'depressao %',
    'saruim': 'Saude ruim %',
    'alcabu': 'consumo abusivo de alcool %'
    
})

# 5. Ordena do maior para o menor % de fumantes e salva em Excel
resultado = resultado.sort_values(by='consumo abusivo de alcool %', ascending=False)

resultado
#resultado.to_excel('dados/vigitel_capitais_18_a_44_anos_sem_peso.xlsx', index=False)
#print("Arquivo 'vigitel_capitais_18_a_44_anos_sem_peso.xlsx' gerado com sucesso!")
```
